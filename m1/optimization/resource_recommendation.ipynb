```python
import pandas as pd

input_file = r"C:\Users\Admin\SlumAI-Optimizer\m1\optimization\resource_allocation.csv"

df = pd.read_csv(input_file)

print("Resource allocation loaded successfully!")
print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData:")
print(df)
```

```python
def get_priority_level(score):

    if score >= 0.70:
        return "Critical"

    elif score >= 0.40:
        return "High"

    elif score >= 0.20:
        return "Medium"

    else:
        return "Low"


df["priority_level"] = df["priority_score"].apply(
    get_priority_level
)
```

```python
def create_recommendation(row):

    if row["allocation_percentage"] < 50:
        return "Increase water allocation"

    elif row["allocation_percentage"] < 80:
        return "Moderate water allocation; monitor demand"

    else:
        return "Water allocation is adequate"


df["recommendation"] = df.apply(
    create_recommendation,
    axis=1
)
```

```python
print("\nFINAL RESOURCE RECOMMENDATIONS")
print("=" * 60)

print(
    df[
        [
            "zone_id",
            "zone_name",
            "predicted_water_demand",
            "risk_score",
            "priority_score",
            "priority_level",
            "optimized_allocation",
            "unmet_demand",
            "allocation_percentage",
            "recommendation"
        ]
    ].sort_values(
        "priority_score",
        ascending=False
    ).to_string(index=False)
)
```

```python
final_recommendations = df[
    [
        "zone_id",
        "zone_name",
        "predicted_water_demand",
        "risk_score",
        "priority_score",
        "priority_level",
        "optimized_allocation",
        "unmet_demand",
        "allocation_percentage",
        "recommendation"
    ]
].sort_values(
    "priority_score",
    ascending=False
)
recommendation_path = (
   "C:\\Users\\Admin\\SlumAI-Optimizer\\m1\\optimization\\resource_recommendations.csv"
)

final_recommendations.to_csv(
    recommendation_path,
    index=False
)

print("\nResource recommendations saved successfully!")
print("Saved to:")
print(recommendation_path)
```
