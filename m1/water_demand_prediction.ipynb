```python
from sklearn.preprocessing import OneHotEncoder

categorical_columns = [
    "zone_id",
    "city",
    "region",
    "notified_status"
]

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

X_train_encoded = encoder.fit_transform(
    X_train[categorical_columns]
)

X_test_encoded = encoder.transform(
    X_test[categorical_columns]
)

print("Encoding completed!")
print("Encoded training shape:", X_train_encoded.shape)
print("Encoded testing shape:", X_test_encoded.shape)
```

```python
# encoding code
# Get numeric columns
numeric_columns = [
    col for col in X_train.columns
    if col not in categorical_columns
]

# Numeric data
X_train_numeric = X_train[numeric_columns]
X_test_numeric = X_test[numeric_columns]

# Combine numeric + encoded data
import numpy as np

X_train_final = np.hstack([
    X_train_numeric.values,
    X_train_encoded
])

X_test_final = np.hstack([
    X_test_numeric.values,
    X_test_encoded
])

print("Final training shape:", X_train_final.shape)
print("Final testing shape:", X_test_final.shape)
```

```python
# Train your first model
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)

model.fit(X_train_final, y_train)

print("Model training completed!")
```

```python
# Make predictions
y_pred = model.predict(X_test_final)

print("Predictions completed!")
print(y_pred[:10])
```

```python
# Check how good the model is
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test, y_pred)

rmse = np.sqrt(
    mean_squared_error(y_test, y_pred)
)

r2 = r2_score(y_test, y_pred)

print("\nModel Evaluation")
print("----------------")
print("MAE :", mae)
print("RMSE:", rmse)
print("R2  :", r2)
```

```python
import pandas as pd
import numpy as np
import joblib

from sqlalchemy import create_engine

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ============================================================
# 1. DATABASE CONNECTION
# ============================================================

engine = create_engine(
    "postgresql://postgres:apoorva2005@localhost:5432/Slum_area_resource_optimizer"
)


# ============================================================
# 2. LOAD DATA
# ============================================================

query = """
SELECT *
FROM feature_engineered_data
ORDER BY date, zone_id;
"""

df = pd.read_sql(query, engine)

print("Rows:", len(df))
print("Columns:", df.columns.tolist())
print(df.head())


# ============================================================
# 3. CONVERT DATE
# ============================================================

df["date"] = pd.to_datetime(df["date"])


# ============================================================
# 4. TIME-BASED TRAIN / TEST SPLIT
# ============================================================

split_date = df["date"].quantile(0.8)

train = df[df["date"] <= split_date].copy()
test = df[df["date"] > split_date].copy()

print("\nSplit date:", split_date)

print("Training rows:", len(train))
print("Testing rows:", len(test))

print("\nTraining dates:")
print(train["date"].min(), "to", train["date"].max())

print("\nTesting dates:")
print(test["date"].min(), "to", test["date"].max())


# ============================================================
# 5. TARGET
# ============================================================

target = "water_demand_liters"


# ============================================================
# 6. FEATURES
# ============================================================

features = [
    "rainfall_mm",
    "water_supply_hours",
    "water_demand_per_capita",
    "waste_per_capita",
    "health_cases_per_1000",
    "population_density_per_km2",
    "people_per_toilet",
    "toilet_water_shortage_pct",
    "nearest_health_center_km",

    # categorical features
    "zone_id",
    "city",
    "region",
    "notified_status"
]


X_train = train[features].copy()
y_train = train[target]

X_test = test[features].copy()
y_test = test[target]


print("\nX_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)


# ============================================================
# 7. CATEGORICAL + NUMERIC FEATURES
# ============================================================

categorical_columns = [
    "zone_id",
    "city",
    "region",
    "notified_status"
]

numeric_columns = [
    "rainfall_mm",
    "water_supply_hours",
    "water_demand_per_capita",
    "waste_per_capita",
    "health_cases_per_1000",
    "population_density_per_km2",
    "people_per_toilet",
    "toilet_water_shortage_pct",
    "nearest_health_center_km"
]


# ============================================================
# 8. PREPROCESSING
# ============================================================

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            categorical_columns
        ),
        (
            "numeric",
            "passthrough",
            numeric_columns
        )
    ]
)


# ============================================================
# 9. LINEAR REGRESSION PIPELINE
# ============================================================

linear_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LinearRegression())
    ]
)


# ============================================================
# 10. TRAIN LINEAR REGRESSION
# ============================================================

linear_model.fit(
    X_train,
    y_train
)

print("\nLinear Regression training completed!")


# ============================================================
# 11. LINEAR REGRESSION PREDICTION
# ============================================================

linear_pred = linear_model.predict(X_test)


# ============================================================
# 12. LINEAR REGRESSION EVALUATION
# ============================================================

linear_mae = mean_absolute_error(
    y_test,
    linear_pred
)

linear_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        linear_pred
    )
)

linear_r2 = r2_score(
    y_test,
    linear_pred
)

print("\nLinear Regression")
print("------------------")
print("MAE :", linear_mae)
print("RMSE:", linear_rmse)
print("R2  :", linear_r2)


# ============================================================
# 13. RANDOM FOREST PIPELINE
# ============================================================

rf_model = Pipeline(
    steps=[
        (
            "preprocessor",
            ColumnTransformer(
                transformers=[
                    (
                        "categorical",
                        OneHotEncoder(
                            handle_unknown="ignore",
                            sparse_output=False
                        ),
                        categorical_columns
                    ),
                    (
                        "numeric",
                        "passthrough",
                        numeric_columns
                    )
                ]
            )
        ),
        (
            "model",
            RandomForestRegressor(
                n_estimators=100,
                random_state=42
            )
        )
    ]
)


rf_model.fit(
    X_train,
    y_train
)

rf_pred = rf_model.predict(
    X_test
)

rf_mae = mean_absolute_error(
    y_test,
    rf_pred
)

rf_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        rf_pred
    )
)

rf_r2 = r2_score(
    y_test,
    rf_pred
)

print("\nRandom Forest")
print("------------------")
print("MAE :", rf_mae)
print("RMSE:", rf_rmse)
print("R2  :", rf_r2)


# ============================================================
# 14. GRADIENT BOOSTING PIPELINE
# ============================================================

gradient_model = Pipeline(
    steps=[
        (
            "preprocessor",
            ColumnTransformer(
                transformers=[
                    (
                        "categorical",
                        OneHotEncoder(
                            handle_unknown="ignore",
                            sparse_output=False
                        ),
                        categorical_columns
                    ),
                    (
                        "numeric",
                        "passthrough",
                        numeric_columns
                    )
                ]
            )
        ),
        (
            "model",
            GradientBoostingRegressor(
                n_estimators=100,
                learning_rate=0.1,
                max_depth=3,
                random_state=42
            )
        )
    ]
)


gradient_model.fit(
    X_train,
    y_train
)

gradient_pred = gradient_model.predict(
    X_test
)

gradient_mae = mean_absolute_error(
    y_test,
    gradient_pred
)

gradient_rmse = np.sqrt(
    mean_squared_error(
        y_test,
        gradient_pred
    )
)

gradient_r2 = r2_score(
    y_test,
    gradient_pred
)

print("\nGradient Boosting")
print("------------------")
print("MAE :", gradient_mae)
print("RMSE:", gradient_rmse)
print("R2  :", gradient_r2)


# ============================================================
# 15. MODEL COMPARISON
# ============================================================

print("\n====================================")
print("MODEL COMPARISON")
print("====================================")

print("\nLinear Regression")
print("MAE :", linear_mae)
print("RMSE:", linear_rmse)
print("R2  :", linear_r2)

print("\nRandom Forest")
print("MAE :", rf_mae)
print("RMSE:", rf_rmse)
print("R2  :", rf_r2)

print("\nGradient Boosting")
print("MAE :", gradient_mae)
print("RMSE:", gradient_rmse)
print("R2  :", gradient_r2)


# ============================================================
# 16. SELECT BEST MODEL
# ============================================================

models = {
    "Linear Regression": {
        "model": linear_model,
        "r2": linear_r2,
        "mae": linear_mae
    },
    "Random Forest": {
        "model": rf_model,
        "r2": rf_r2,
        "mae": rf_mae
    },
    "Gradient Boosting": {
        "model": gradient_model,
        "r2": gradient_r2,
        "mae": gradient_mae
    }
}


# Highest R2 = better
# If R2 is similar, lower MAE is better

best_model_name = max(
    models,
    key=lambda name: (
        models[name]["r2"],
        -models[name]["mae"]
    )
)

best_model = models[best_model_name]["model"]

print("\n====================================")
print("BEST MODEL")
print("====================================")

print("Best model:", best_model_name)


# ============================================================
# 17. SAVE COMPLETE PIPELINE
# ============================================================

model_path = (
    r"C:\Users\Admin\SlumAI-Optimizer\models\water_demand_model.pkl"
)

joblib.dump(
    best_model,
    model_path
)

print("\nBest model saved successfully!")
print(model_path)


# ============================================================
# 18. SAVE PREDICTIONS
# ============================================================

best_prediction = models[
    best_model_name
]["model"].predict(X_test)


results = test[
    [
        "date",
        "zone_id",
        "zone_name",
        "water_demand_liters"
    ]
].copy()

results["predicted_water_demand"] = best_prediction


output_path = (
    r"C:\Users\Admin\SlumAI-Optimizer\outputs\predictions\water_demand_predictions.csv"
)

results.to_csv(
    output_path,
    index=False
)

print("\nPredictions saved successfully!")
print(output_path)
```
